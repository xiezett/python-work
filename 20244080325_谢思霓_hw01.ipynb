```python
def gcd(m, n):
    while n != 0:
        temp = m % n
        m = n
        n = temp
    return m

print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))

print("负数测试 gcd(-6, 3):", gcd(-6, 3))
```

```python
import math

def euler_limit(n):
    return (1 + 1 / n) ** n

def euler_infinite_sum(n):
    total = 0.0
    fact = 1
    for k in range(n):
        if k == 0:
            fact = 1
        else:
            fact = fact * k
        total += 1.0 / fact
    return total

def euler_approx(epsilon):
    total = 0.0
    fact = 1
    k = 0
    while True:
        term = 1.0 / fact
        if term < epsilon:
            break
        total += term
        k += 1
        fact = fact * k
    return total

def print_euler_sum_table(n):
    for k in range(1, n + 1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")

print("euler_limit(1000):", euler_limit(1000))
print("euler_infinite_sum(4):", euler_infinite_sum(4))
print("euler_approx(0.001):", euler_approx(0.001))
print_euler_sum_table(4)
print("math.exp(1)真实e：", math.exp(1))
```

```python
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

def is_power(b, n):
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True

print(is_power_of_2(4))
print(is_power_of_2(5))
print(is_power(6, 36))
print(is_power(7, 12))
```
