````python
import re

text1 = "cat on mat; later: cats and dogs"
print("数字 \\d+ :", re.findall(r"\d+", text1))
print("空白 \\s+ :", re.findall(r"\s+", text1))
print("单词 cat \\bcat\\b :", re.findall(r"\bcat\b", text1))

text2 = "cat dog bird CAT"
print("\ncat|dog :", re.findall(r"cat|dog", text2))

print("a|aa|aaa on 'aaaa' :", re.findall(r"a|aa|aaa", "aaaa"))
print("aaa|aa|a on 'aaaa' :", re.findall(r"aaa|aa|a", "aaaa"))

text3 = "ct cat caat caaat"
for pat in [r"ca*t", r"ca+t", r"ca?t", r"ca{2}t", r"ca{1,2}t"]:
    print(f"{pat:10s} ->", re.findall(pat, text3))
````

```text
任务 2 解释：
cat|dog 匹配 cat 或 dog。由于正则默认区分大小写，因此不匹配大写的 CAT。
a|aa|aaa 引擎从左到右优先匹配，先匹配到单个 a 就结束，结果为 ['a', 'a', 'a', 'a']。
aaa|aa|a 长选项排在前，优先匹配最长项，结果为 ['aaa', 'a']。
任务 3 解析（量词含义）：
ca*t 星号表示 a 出现 0 次或多次。匹配结果: ct, cat, caat, caaat。
ca+t 加号表示 a 出现 1 次或多次。匹配结果: cat, caat, caaat (不含 ct)。
ca?t 问号表示 a 出现 0 次或 1 次。匹配结果: ct, cat。
ca{2}t {2} 表示 a 恰好出现 2 次。匹配结果: caat。
ca{1,2}t {1,2} 表示 a 出现 1 到 2 次。匹配结果: cat, caat。
```

````python
import re

text_email = "联系: alice@example.com; 备选: bob@test.org"
emails = re.findall(r"([\w.+-]+)@([\w.-]+)", text_email)
for user, domain in emails:
    print(f"用户名: {user:8s} 域名: {domain}")

text_dup = "cat cat, dog cat, bird bird"
dups = re.findall(r"\b(\w+)\s+\1\b", text_dup)
print("\n连续重复单词:", dups)

text_date = "2024-05-01\n2024/05/02\n2024.05.03"
print("\n转换前:\n" + text_date)
converted = re.sub(r"(\d{4})[-/.](\d{2})[-/.](\d{2})", r"\2-\3-\1", text_date)
print("转换后:\n" + converted)

text_phone = "联系我: 123-456-7890 或 987.654.3210"
unified = re.sub(r"(\d{3})[-.](\d{3})[-.](\d{4})", r"(\1) \2-\3", text_phone)
print("转换后:", unified)
````

````python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲: 正则表达式</p>
<p class="course">第七讲: 结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")

print("页面标题:", soup.title.get_text())
h1 = soup.find("h1")
print("h1 标签名:", h1.name, " 文本:", h1.get_text())

print("\nclass=course 段落")
for p in soup.find_all("p", class_="course"):
    print(p.get_text())

print("\n所有链接")
for a in soup.find_all("a"):
    print(f"文字: {a.get_text()}  href: {a['href']}")

print("\nget_text() 对比")
print("整个文档:", repr(soup.get_text()))
print("单个段落:", repr(soup.find('p', class_='course').get_text()))
print("单个链接:", repr(soup.find('a').get_text()))
````

```text
任务 3 结果解析（get_text() 文本提取对比）：
整个文档 (soup.get_text())：
结果提取了 HTML 中所有层级的纯文本。由于原始 HTML 源码中标签之间存在换行和缩进，提取出的字符串包含了多个换行符 \n 和空白符，把所有内容拼接在了一起。这适合需要全文搜索或整页内容分析的场景。
单个段落 (soup.find('p', class_='course').get_text())：
结果精准提取了指定 <p> 标签内部的文字。去除了 HTML 结构，只保留了干净的本段文字，没有任何多余的换行符。
单个链接 (soup.find('a').get_text())：
结果仅提取了 <a> 标签内的文字，不包含 href 属性中的网址内容。这说明 get_text() 只关心标签包裹的展示文本。
```
