```python
def is_palindrome(s):
    temp = s.replace(" ","").lower()
    return temp == temp[::-1]


def is_abecedarian(s):
    temp = s.replace(" ","").lower()
    for i in range(len(temp)-1):
        if temp[i] > temp[i+1]:
            return False
    return True


def remove_vowels(s):
    vowel = ['a','e','i','o','u','A','E','I','O','U']
    new_s = ""
    for c in s:
        if c not in vowel:
            new_s = new_s + c
    return new_s


print("is_palindrome('T A C O cat'):", is_palindrome("T A C O cat"))
print("is_palindrome('level'):", is_palindrome("level"))
print("is_abecedarian('abcd efgh xyz'):", is_abecedarian("abcd efgh xyz"))
print("remove_vowels('cat'):", remove_vowels("cat"))
print("remove_vowels('audio'):", remove_vowels("audio"))
```

```python
def list_reverse(lst):
    if type(lst) != list:
        raise TypeError("输入必须是列表")
    res = lst[::-1]
    return res


def is_sorted(seq):
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True


print("list_reverse([1,2,3]):", list_reverse([1,2,3]))
print("is_sorted([1,2,3,4]):", is_sorted([1,2,3,4]))
print("is_sorted([1,5,3]):", is_sorted([1,5,3]))
```

```python
def vec_scalar_mult(t, s):
    if not isinstance(t,tuple):
        raise TypeError("第一个参数需要是元组")
    if not isinstance(s,(int,float)):
        raise TypeError("标量只能是数字")
    for num in t:
        if not isinstance(num,(int,float)):
            raise TypeError("向量元素只能是数字")
    result = tuple(x*s for x in t)
    return result


def vec_inner_product(x,y):
    if not isinstance(x,tuple) or not isinstance(y,tuple):
        raise TypeError("输入必须是元组")
    if len(x) != len(y):
        raise ValueError("两个向量长度不一样")
    sum_total = 0.0
    for a,b in zip(x,y):
        sum_total += a*b
    return sum_total


def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return True
    row_len = len(mx[0])
    for row in mx:
        if not isinstance(row,tuple):
            return False
        if len(row) != row_len:
            return False
        for n in row:
            if not isinstance(n,(int,float)):
                return False
    return True


def mx_vec_mult(mx, vec):
    if check_valid_mx(mx) == False:
        raise TypeError("不是合法矩阵")
    if not isinstance(vec,tuple):
        raise TypeError("向量需要是元组")
    if len(mx[0]) != len(vec):
        raise ValueError("矩阵列数和向量不匹配")
    out_list = []
    for r in mx:
        tmp = 0.0
        for a,b in zip(r,vec):
            tmp = tmp + a*b
        out_list.append(tmp)
    return tuple(out_list)


print("vec_scalar_mult((1,2,3),2):", vec_scalar_mult((1,2,3),2))
print("vec_inner_product((1,2),(3,4)):", vec_inner_product((1,2),(3,4)))
print("check_valid_mx(((1,2),(3,4))):", check_valid_mx(((1,2),(3,4))))
print("mx_vec_mult(((1,2),(3,4)), (5,6)):", mx_vec_mult(((1,2),(3,4)), (5,6)))
```

```python
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim,int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能是负数")
        if elements is None:
            self.elements = [0.0]*dim
        else:
            if not isinstance(elements,(list,tuple)):
                raise TypeError("elements只能是列表或者元组")
            if len(elements) != dim:
                raise ValueError("元素个数和维度对不上")
            self.elements = list(elements)
        self.dim = dim

    def __eq__(self,other):
        if not isinstance(other,Vector):
            raise TypeError("只能和Vector比较")
        if self.dim != other.dim:
            raise ValueError("维度不一样不能比较")
        return self.elements == other.elements

    def __lt__(self,other):
        if not isinstance(other,Vector):
            raise TypeError("只能和Vector比较")
        if self.dim != other.dim:
            raise ValueError("维度不一样不能比较")
        return self.elements < other.elements

    def __le__(self,other):
        return self < other or self == other

    def __gt__(self,other):
        return not self <= other

    def dot(self,other):
        if not isinstance(other,Vector):
            raise TypeError("参数必须是Vector实例")
        if self.dim != other.dim:
            raise ValueError("维度不一致，无法计算内积")
        res = 0.0
        for a,b in zip(self.elements, other.elements):
            res += a*b
        return res

    def __mul__(self,rhs):
        if isinstance(rhs,(int,float)):
            new_e = [x*rhs for x in self.elements]
            return Vector(self.dim, new_e)
        elif isinstance(rhs,Vector):
            if self.dim != rhs.dim:
                raise ValueError("向量维度不一致")
            new_e = []
            for a,b in zip(self.elements,rhs.elements):
                new_e.append(a*b)
            return Vector(self.dim, new_e)
        else:
            raise TypeError("不支持这个类型相乘")

    def __rmul__(self,lhs):
        if isinstance(lhs,(int,float)):
            return self * lhs
        else:
            raise TypeError("不支持这个类型相乘")

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"


v1 = Vector(2,(2,3))
v2 = Vector(2,(4,5))
print("v1 =", v1)
print("v1.dot(v2):", v1.dot(v2))
print("v1 * v2:", v1 * v2)
print("2 * v1:", 2 * v1)
print("v1 * 3:", v1 * 3)
print("v1 < v2:", v1 < v2)
```
