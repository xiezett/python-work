```python
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须在0‑100之间")
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

sum_for = 0
for i in range(1, 101):
    sum_for += i
sum_while = 0
n = 1
while n <= 100:
    sum_while += n
    n += 1
print("for循环1‑100和：", sum_for)
print("while循环1‑100和：", sum_while)

def fact_iter(n):
    res = 1
    for i in range(2, n+1):
        res *= i
    return res

def fact_recur(n):
    if n == 0:
        return 1
    return n * fact_recur(n-1)

for num in [0,1,5,10]:
    print(f"n={num},迭代={fact_iter(num)},递归={fact_recur(num)}")

def summary(values):
    mini = min(values)
    maxi = max(values)
    avge = sum(values)/len(values)
    return (mini, maxi, avge)

res = summary([0,1,5,10])
mi,ma,av = res
print("summary元组解包：", mi,ma,av)

def sum_all(*args):
    return sum(args)

print("无参sum_all():", sum_all())
print("单参sum_all(5):", sum_all(5))
print("多参sum_all(1,2,3,4):", sum_all(1,2,3,4))
```

```python
import os
import pickle

info_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

print("当前工作路径：", os.getcwd())
save_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(save_dir, exist_ok=True)
stu_txt = os.path.join(save_dir, "students.txt")

with open(stu_txt, "w", encoding="utf-8") as fp:
    fp.write(info_text)

with open(stu_txt, "r", encoding="utf-8") as fp:
    all_lines = fp.readlines()


def parse_line(single_line):
    single_line = single_line.strip()
    if len(single_line) == 0:
        raise ValueError("这一行是空的")
    res = single_line.split(",")
    if len(res) != 2:
        raise ValueError("一行里面的数据项不是两个")
    s_name, s_score_str = res
    s_name = s_name.strip()
    s_score_str = s_score_str.strip()
    if s_name == "":
        raise ValueError("姓名为空")
    s_score = int(s_score_str)
    if s_score < 0 or s_score > 100:
        raise ValueError("分数必须在0到100之间")
    return {"name": s_name, "score": s_score}


good_list = []
fail_list = []
for line_no, one_line in enumerate(all_lines, start=1):
    if line_no == 1:
        continue
    try:
        item = parse_line(one_line)
        good_list.append(item)
    except Exception as msg:
        fail_list.append((line_no, str(msg)))

print("解析成功的数据：", good_list)
print("解析失败（行号，原因）：", fail_list)


count_good = len(good_list)
count_fail = len(fail_list)
avg = 0.0
pass_rate = 0.0
if count_good > 0:
    score_sum = 0
    for data in good_list:
        score_sum += data["score"]
    avg = score_sum / count_good

    pass_num = 0
    for data in good_list:
        if data["score"] >= 60:
            pass_num += 1
    pass_rate = pass_num / count_good

report_txt = os.path.join(save_dir, "report.txt")
with open(report_txt, "w", encoding="utf-8") as fp:
    fp.write(f"有效记录数量：{count_good}\n")
    fp.write(f"无效记录数量：{count_fail}\n")
    fp.write(f"平均分：{avg:.2f}\n")
    fp.write(f"及格率：{pass_rate:.1%}\n")

with open(report_txt, "a", encoding="utf-8") as fp:
    fp.write("加进去的测试文字:数据处理作业\n")


pickle_path = os.path.join(save_dir, "students.pkl")
with open(pickle_path, "wb") as fp:
    pickle.dump(good_list, fp)

with open(pickle_path, "rb") as fp:
    load_data = pickle.load(fp)
print("pickle读取的数据和原始数据是否一样：", load_data == good_list)


module_file = os.path.join(os.getcwd(), "student_utils.py")
func_code = '''
def parse_line(single_line):
    single_line = single_line.strip()
    if len(single_line) == 0:
        raise ValueError("这一行是空的")
    res = single_line.split(",")
    if len(res) != 2:
        raise ValueError("一行里面的数据项不是两个")
    s_name, s_score_str = res
    s_name = s_name.strip()
    s_score_str = s_score_str.strip()
    if s_name == "":
        raise ValueError("姓名为空")
    s_score = int(s_score_str)
    if s_score < 0 or s_score > 100:
        raise ValueError("分数必须在0到100之间")
    return {"name": s_name, "score": s_score}
'''
with open(module_file, "w", encoding="utf-8") as fp:
    fp.write(func_code)

from student_utils import parse_line
print("模块测试，正常数据：", parse_line("Tom,77"))
try:
    parse_line("BadData")
except Exception as e:
    print("模块测试，错误数据：", e)


try:
    open("not_exist_file.txt", "r")
except FileNotFoundError as e:
    print("捕获到文件找不到的异常：", e)
```

```python
import copy

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def isPassed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"


stuData = [
    {'name': 'Alice', 'score': 85},
    {'name': 'Bob', 'score': 58},
    {'name': 'Carol', 'score': 92},
    {'name': 'David', 'score': 76},
    {'name': 'Eva', 'score': 60},
    {'name': 'Frank', 'score': 89}
]

stuList = []
for item in stuData:
    temp = Student(item["name"], item["score"])
    stuList.append(temp)
    print(temp, "及格状态:", temp.isPassed())


class GraduateStudent(Student):
    def __init__(self, name, score, topic):
        super().__init__(name, score)
        self.topic = topic

    def __str__(self):
        return f"GraduateStudent(name={self.name},score={self.score},topic={self.topic})"

gStu = GraduateStudent("LiHua",82,"AI")
print(gStu, "及格：", gStu.isPassed())


base = {"name": "Alice", "scores": [85, 90]}
o1 = base
o1["scores"][0] = 99
print("直接赋值后base:", base)

base = {"name": "Alice", "scores": [85, 90]}
o2 = copy.copy(base)
o2["scores"][0] = 99
print("浅拷贝后base:", base)

base = {"name": "Alice", "scores": [85, 90]}
o3 = copy.deepcopy(base)
o3["scores"][0] = 99
print("深拷贝后base:", base)
```
